# Treinamento

## Primeira Etapa
- Carregar o torch
- Converter a imagem para tensor
- Carregar o dataset de treinamento (MNIST)
- Iterar sobre os dados e construir os batchs

```python
# A primeira célula (ChT7GVR7bwFh) importa as bibliotecas torch e torchvision,
# define uma transformação para converter as imagens para tensor, baixa e carrega
# o conjunto de dados MNIST para treinamento (tr_data), cria um DataLoader para
# iterar sobre os dados em lotes (tr_loader) e imprime o tamanho do conjunto de dados de treinamento.


import torch
from torchvision import datasets, transforms

transform = transforms.ToTensor()
tr_data = datasets.MNIST(root = './dataset', train = True, \
                         download = True, transform = transform)

tr_loader = torch.utils.data.DataLoader(tr_data, batch_size = 10, \
                                        shuffle = True)
tam_dataset = len(tr_data)
print(f'Tamanho: {tam_dataset}')
```

#### Exemplo de dado do dataset MNIST

```python
# A segunda célula (UAnytswpbzVZ) importa a biblioteca matplotlib para plotagem,
# seleciona a primeira imagem e seu rótulo do conjunto de dados de treinamento,
#exibe a imagem em escala de cinza e define o título do plot com o rótulo correspondente.

from matplotlib import pyplot as plt

img, label = tr_data[0]
plt.imshow(img.squeeze(), cmap = 'gray')
plt.title(f'Label: {label}')
plt.show()
```

## Segunda Etapa
Construir a arquitetura da rede
- Entrada
- Número de classes
- Modelo Sequencia
- Uma camada linear

```python
# A terceira célula (-01hOQsnb5yA) importa módulos do torch.nn para construir a arquitetura do modelo,
# define o tamanho de entrada e o número de classes, e cria um modelo sequencial com uma camada linear
# de entrada para 10 classes (saída). Finalmente, imprime a estrutura do modelo.

from torch.nn import Sequential, Linear, ReLU

tam_entrada = 28*28
n_classes = 10

# Modelo com apenas uma camada linear
# model = Sequential(Linear(tam_entrada, 128), ReLU(), Linear(128, 10))
model = Sequential(Linear(tam_entrada, n_classes))
print(model)
```

## Terceira Etapa
Treinamento do modelo

- com função de perda e otimizador definidos
- faz o loop de treinamento pelas epocas definidas
- atualiza pesos com *retropropagation*
- calcula perda e atualiza pesos do modelo

```python
# A quarta célula (wVxDuM6gb865) importa torch e tqdm, define a função de perda (CrossEntropyLoss) e o otimizador (SGD),
# define o número de épocas e inicializa uma lista para armazenar as perdas. Em seguida,
# inicia um loop para treinar o modelo por um número definido de épocas, iterando sobre os lotes de
# dados de treinamento, calculando a perda, realizando a retropropagação e atualizando os
# pesos do modelo.


import torch
from tqdm import tqdm

criterion = torch.nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr = 0.01)
epochs = 5
losses = []
for epoch in range(epochs):
    print(f'Epoch: {epoch}')
    batch_loss = 0
    #for img_batch, label_batch in tr_loader:
    for img_batch, label_batch in tqdm(tr_loader):
        img_batch = img_batch.view(img_batch.size(0), -1)
        optimizer.zero_grad()
        output = model(img_batch)
        loss = criterion(output, label_batch)
        #losses.append(loss.item())
        batch_loss += loss.item()
        loss.backward()
        optimizer.step()
    losses.append(batch_loss/len(tr_loader))
```

```python
# O print(losses) irá exibir uma lista de números. Cada número nessa lista representa a
# média da função de perda calculada ao final de cada época de treinamento.
# Como você definiu epochs = 5 na célula de treinamento (wVxDuM6gb865), você verá 5 números na lista losses,
# um para cada época. O valor da perda geralmente diminui ao longo das épocas, indicando que o modelo está aprendendo a fazer previsões melhores.

print(losses)

# [0.5347098572583249, 0.35873828949096304, 0.3306599416344737, 0.3162760896642382, 0.30684097594651394]
```

# Analise Importancia Pixels
Selecionar quais são os pixels mais importantes para a saída, para poder diminuir o modelo

```python
import torch
import matplotlib.pyplot as plt


model.eval()
# Get the weights from the first linear layer
# Calculate pixel importance (sum of absolute weights across output classes)
with torch.no_grad():
  weights = model[0].weight.data
  pixel_importance = torch.sum(torch.abs(weights), dim=0).reshape(28, 28)

```

```python
plt.imshow(pixel_importance, cmap='hot')
plt.title("Mapa de Importância dos Pixels (MLP Linear)")
plt.colorbar()
plt.show()
```

```python
for target_label in range(10):
    # encontra a primeira imagem com o label desejado
    for img, label in tr_data:
        if label == target_label:
            break  # img e label encontrados

    # normaliza a importância dos pixels
    pixel_importance_norm = (pixel_importance - pixel_importance.min()) / \
                            (pixel_importance.max() - pixel_importance.min())

    # plota
    plt.figure(figsize=(5,5))
    plt.imshow(img.squeeze(), cmap='gray')
    plt.imshow(pixel_importance_norm, cmap='hot', alpha=0.5)
    plt.title(f'Label: {label}')
    plt.colorbar(label='Importância do pixel')
    plt.axis('off')
    plt.show()
```

```python


num_top_pixels = new_dimension * new_dimension
# Flatten the 28x28 pixel_importance tensor to get a 1D tensor
pixel_importance_flat = pixel_importance.flatten()
# Sort the flattened tensor in descending order and get the indices
top_pixel_indices = torch.argsort(pixel_importance_flat, descending=True)[:num_top_pixels]
# Convert the flat indices back to 2D coordinates (row, col)
# Index i in the flattened tensor corresponds to row = i // 28, col = i % 28
top_pixel_coordinates = [(i // 28, i % 28) for i in top_pixel_indices]
```

```python

print(f"Top {num_top_pixels} most important pixel coordinates (row, col):")
for coord in top_pixel_coordinates:
    print(coord)
```

```python
# A quinta célula (nof7Z6R1cEaz) importa datasets, transforms e DataLoader do torchvision e torch.utils.data.
# Ela baixa e carrega o conjunto de dados MNIST para teste (te_data) e cria um DataLoader para iterar sobre os dados de teste em lotes (te_loader).

from torchvision import datasets, transforms
from torch.utils.data import DataLoader

te_data = datasets.MNIST(root = './dataset', train = False, transform = transforms.ToTensor())

te_loader = DataLoader(te_data, batch_size = 10, shuffle = False)
```

## Segunda Etapa
- desativa camadas que não são usadas no teste
- desativa o calculo de gradiente (pois pesos não serão mais atualizados)
- passa a imagem na rede totalmente conectada
- calcula loss (quão distante predições estão dos rótulos reais)
- retorna média de perdas por batch

### Calculo de Loss

```python
# A sexta célula (MIJUYZY9cMsm) define a função calc_loss que calcula a perda média em um conjunto de dados usando o
# modelo e a função de perda fornecidos. Ela coloca o modelo em modo de avaliação (model.eval()) e desativa o cálculo
# de gradientes (torch.no_grad()) para economizar memória e computação durante a avaliação.

def calc_loss(model, criterion, loader):
    batch_losses = []
    model.eval()
    with torch.no_grad():
        for img_batch, label_batch in loader:
            img_batch = img_batch.view(img_batch.size(0), -1)
            output = model(img_batch)
            loss = criterion(output, label_batch).item()  # <- item() converte para float
            batch_losses.append(loss)

    mean_loss = sum(batch_losses) / len(batch_losses)
    return batch_losses, mean_loss
```

```python
# A sétima célula (YvEqW--FcRVk) chama a função calc_loss para calcular a perda no conjunto de
# dados de teste usando o modelo treinado e a função de perda definida anteriormente, e imprime a perda calculada.
batch_losses, mean_loss = calc_loss(model, criterion, te_loader)
print(f"Loss média: {mean_loss:.4f}")
print(f"Loss por batch (primeiros 5): {batch_losses[:10]}")

```

```python
# A oitava célula (VOIPL7kkccX2) importa numpy, seleciona a primeira imagem e seu
# rótulo do conjunto de dados de teste, passa a imagem pelo modelo para obter uma
# previsão e imprime o rótulo real, a saída do modelo (logits) e o rótulo previsto (o índice com o maior logit).

import numpy as np
img, label = te_data[0]
pred = model(img.view(1, -1))
print(f'Label: {label}')
print(f'Prediction: {pred}')
print(f"Label predicted: {pred.argmax()}")
```

### Calculo da Acurácia

```python
# v1 - usando Torch
# A nona célula (RfOKRtY0cfoY) define a função calc_accuracy que calcula a acurácia
# do modelo em um conjunto de dados usando torch. Ela itera sobre os lotes de dados,
# compara as previsões do modelo com os rótulos reais e retorna a proporção de previsões corretas.
def calc_accuracy(model, criterion, loader, n_classes=10):
    correct_total = 0
    total_total = 0
    correct_per_class = {i: 0 for i in range(n_classes)}
    total_per_class = {i: 0 for i in range(n_classes)}

    model.eval()
    with torch.no_grad():
        for img_batch, label_batch in loader:
            img_batch = img_batch.view(img_batch.size(0), -1)
            output = model(img_batch)

            preds = output.argmax(dim=1)
            correct_total += (preds == label_batch).sum().item()
            total_total += len(label_batch)

            # Acertos por classe
            for i in range(len(label_batch)):
                label = label_batch[i].item()
                total_per_class[label] += 1
                if preds[i] == label_batch[i]:
                    correct_per_class[label] += 1

    # Calcula acurácia por classe
    class_acc = {cls: correct_per_class[cls]/total_per_class[cls] if total_per_class[cls] > 0 else 0
                 for cls in range(n_classes)}

    overall_acc = correct_total / total_total
    return overall_acc, class_acc
```

```python
overall_acc, class_acc = calc_accuracy(model, criterion, te_loader)

print(f"Acurácia geral: {overall_acc*100:.2f}%")
for cls, acc in class_acc.items():
    print(f"Classe {cls}: {acc*100:.2f}%")
```

# Treinamento Reduzido


Função que reduz imagem original


```python
def reduce_image(img_tensor, coords):
    """
    Recebe uma imagem [1, 28, 28] e devolve tensor [num_top_pixels]
    contendo apenas os pixels mais importantes.
    """
    img = img_tensor.view(28, 28)
    return torch.tensor([img[r, c].item() for (r, c) in coords])
```

## Primeira Etapa
- Criar dataset MNIST reduzido
- Já gera o dataset como vetor 1D

```python
from torch.utils.data import TensorDataset, DataLoader

# Reduz conjunto de treino
X_train_reduced = torch.stack([reduce_image(img, top_pixel_coordinates) for img, _ in tr_data])
y_train = torch.tensor([label for _, label in tr_data])

# Reduz conjunto de teste
X_test_reduced = torch.stack([reduce_image(img, top_pixel_coordinates) for img, _ in te_data])
y_test = torch.tensor([label for _, label in te_data])

print(f"Novo shape de treino: {X_train_reduced.shape}")
print(f"Novo shape de teste:  {X_test_reduced.shape}")

# DataLoaders
batch_size = 10
train_loader_reduced = DataLoader(TensorDataset(X_train_reduced, y_train), batch_size=batch_size, shuffle=True)
test_loader_reduced = DataLoader(TensorDataset(X_test_reduced, y_test), batch_size=batch_size)
```

## Segunda etapa
- criar modelo reduzido


```python
import torch.nn as nn

num_pixels = new_dimension * new_dimension
model_small = nn.Sequential(
    nn.Linear(num_pixels, 10)
)
```

## Terceira etapa
- realizar treinamento

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model_small.parameters(), lr=0.1)

epochs = 5
losses_reduced = []
for epoch in range(epochs):
    print(f'Epoch: {epoch}')
    model_small.train()
    total_loss = 0
    for X_batch, y_batch in tqdm(train_loader_reduced):
        optimizer.zero_grad()
        output = model_small(X_batch)
        loss = criterion(output, y_batch)
        total_loss += loss.item()
        loss.backward()
        optimizer.step()
    losses_reduced.append(total_loss/len(train_loader_reduced))
    #print(f"Epoch {epoch+1}, Loss: {total_loss/len(train_loader_reduced):.4f}")
```

## Primeira Etapa
Carregar dataset para teste reduzido

```python
from torch.utils.data import TensorDataset, DataLoader

# Lista de imagens reduzidas
X_test_reduced = torch.stack([reduce_image(img, top_pixel_coordinates) for img, _ in te_data])
y_test = torch.tensor([label for _, label in te_data])

print(f"Shape do dataset reduzido: {X_test_reduced.shape}")  # deve ser (10000, 196)

# DataLoader para o modelo reduzido
test_loader_reduced = DataLoader(
    TensorDataset(X_test_reduced, y_test),
    batch_size=10,  # mantém mesmo batch size do modelo normal
    shuffle=False
)
```

## Segunda etapa
- desativa camadas que não são usadas no teste
- desativa o calculo de gradiente (pois pesos não serão mais atualizados)
- passa a imagem na rede totalmente conectada
- calcula loss (quão distante predições estão dos rótulos reais)
- retorna média de perdas por batch

```python
def calc_loss_reduced(model, criterion, loader):
    """
    Calcula a loss média de um modelo MLP que recebe entradas já achatadas.
    loader: DataLoader do dataset reduzido (ex: test_loader_reduced)
    """
    batch_losses = []
    model.eval()
    with torch.no_grad():
        for img_batch, label_batch in loader:
            # img_batch já está achatado, não precisa de .view()
            output = model(img_batch)
            loss = criterion(output, label_batch).item()
            batch_losses.append(loss)

    mean_loss = sum(batch_losses) / len(batch_losses)
    return batch_losses, mean_loss
```

```python
batch_losses_reduced, mean_loss_reduced = calc_loss_reduced(model_small, criterion, test_loader_reduced)

print(f"Loss média: {mean_loss_reduced:.4f}")
print(f"Loss por batch (primeiros 5): {batch_losses_reduced[:10]}")
```

```python
# Pega a primeira imagem do dataset reduzido
img, label = X_test_reduced[0], y_test[0]

# img já tem shape (196,), então só precisamos adicionar dimensão de batch
pred = model_small(img.unsqueeze(0))  # shape -> (1, 196)

print(f"Label real: {label.item()}")
print(f"Prediction (raw logits): {pred}")
print(f"Label previsto: {pred.argmax().item()}")
```

```python
def calc_accuracy_reduced(model, criterion, loader, n_classes=10):
    """
    Calcula a acurácia geral e a acurácia por classe de um modelo MLP.

    Retorna:
        overall_acc: acurácia total
        class_acc: dicionário {classe: acurácia}
    """
    correct_total = 0
    total_total = 0
    correct_per_class = {i: 0 for i in range(n_classes)}
    total_per_class = {i: 0 for i in range(n_classes)}

    model.eval()
    with torch.no_grad():
        for img_batch, label_batch in loader:
            # img_batch já está achatado, não precisa de .view()
            output = model(img_batch)
            preds = output.argmax(dim=1)

            # Acertos gerais
            correct_total += (preds == label_batch).sum().item()
            total_total += len(label_batch)

            # Acertos por classe
            for i in range(len(label_batch)):
                label = label_batch[i].item()
                total_per_class[label] += 1
                if preds[i] == label_batch[i]:
                    correct_per_class[label] += 1

    # Calcula acurácia por classe
    class_acc = {cls: correct_per_class[cls]/total_per_class[cls] if total_per_class[cls] > 0 else 0
                 for cls in range(n_classes)}

    overall_acc = correct_total / total_total
    return overall_acc, class_acc
```

```python
overall_acc_reduced, class_acc_reduced = calc_accuracy_reduced(model_small, criterion, test_loader_reduced)

print(f"Acurácia geral: {overall_acc_reduced*100:.2f}%")
for cls, acc in class_acc_reduced.items():
    print(f"Classe {cls}: {acc*100:.2f}%")
```

```python
import torch.nn as nn

num_pixels = new_dimension * new_dimension
model_random = nn.Sequential(
    nn.Linear(num_pixels, 10)
)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model_random.parameters(), lr=0.1)

epochs = 5
losses_random = []
for epoch in range(epochs):
    print(f'Epoch: {epoch}')
    model_random.train()
    total_loss = 0
    for X_batch, y_batch in tqdm(train_loader_random):
        optimizer.zero_grad()
        output = model_random(X_batch)
        loss = criterion(output, y_batch)
        total_loss += loss.item()
        loss.backward()
        optimizer.step()
    losses_random.append(total_loss/len(train_loader_reduced))
    #print(f"Epoch {epoch+1}, Loss: {total_loss/len(train_loader_reduced):.4f}")
```
